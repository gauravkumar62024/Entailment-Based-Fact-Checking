```python
# Install dependencies (if not already installed)
# pip install bertopic[all]

from bertopic import BERTopic
from sklearn.feature_extraction.text import CountVectorizer

# Load dataset from file (assuming one claim per line)
with open("claims_only.txt", "r") as file:
    documents = [line.strip() for line in file if line.strip()]

# Initialize BERTopic with custom vectorizer
vectorizer_model = CountVectorizer(stop_words="english")
topic_model = BERTopic(vectorizer_model=vectorizer_model, verbose=True)

# Fit BERTopic model
topics, probs = topic_model.fit_transform(documents)

# Print summary of topics
print("\nTop Topics Summary:")
print(topic_model.get_topic_info().head(10))  # Top 10 topics

# Print representative words for each topic
print("\nTopics and their top words:\n")
for topic_num in range(len(set(topics)) - 1):  # -1 to exclude -1 (outlier)
    print(f"Topic {topic_num}:")
    print(topic_model.get_topic(topic_num))  # List of (word, weight)
    print()

# Optional: print documents for a topic
topic_id = 0  # Change this to see other topics
print(f"\nSample documents from Topic {topic_id}:\n")
docs_for_topic = [doc for doc, t in zip(documents, topics) if t == topic_id]
for i, doc in enumerate(docs_for_topic[:5]):  # Limit to 5 docs
    print(f"- {doc}")
```
